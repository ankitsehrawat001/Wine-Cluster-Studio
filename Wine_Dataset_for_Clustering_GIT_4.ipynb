## Case Study: Data Acquisition

This cell uses the `kagglehub` library to download a dataset from Kaggle. The `dataset_download` function retrieves the specified dataset ('harrywang/wine-dataset-for-clustering') and stores its local path in the `path` variable. This is a common first step in data science projects to obtain the necessary data.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("harrywang/wine-dataset-for-clustering")

print("Path to dataset files:", path)
```

## Case Study: Library Imports

This cell imports all the necessary Python libraries for data manipulation, analysis, and visualization. It includes `pandas` for data handling, `numpy` for numerical operations, `matplotlib.pyplot` and `seaborn` for plotting, `StandardScaler` for data preprocessing, `KMeans` for clustering, and `silhouette_score` for evaluating clustering performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## Case Study: Data Loading

This cell loads the downloaded CSV file into a pandas DataFrame named `df`. It constructs the full file path using `os.path.join` and the `path` variable obtained from the previous data acquisition step. This is a crucial step to get the data ready for further exploration and processing.

```python
import pandas as pd
import os

df = pd.read_csv(os.path.join(path, 'wine-clustering.csv'))
```

## Case Study: Displaying DataFrame

This simple cell displays the entire `df` DataFrame. This allows for a quick inspection of the loaded data, including column names, data types, and initial rows, to ensure the data was loaded correctly and to get a preliminary understanding of its structure.

```python
df
```

## Case Study: Feature Scaling

This cell performs feature scaling on the entire dataset. It uses `StandardScaler` from `sklearn.preprocessing` to transform the features such that they have a mean of 0 and a standard deviation of 1. This step is essential for many machine learning algorithms, especially distance-based ones like K-Means, to prevent features with larger ranges from dominating the distance calculations. The scaled data is stored in `X_scaled`.

```python
# Saare features numeric hain, isliye poore data ko scale karenge
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Check karne ke liye
print("Saare features ab 0 se 1 ke bich ke scale par aa gaye hain.")
```

## Case Study: Elbow Method for Optimal K

This cell implements the Elbow Method to determine the optimal number of clusters (K) for K-Means. It iterates from 1 to 10 clusters, fits a K-Means model to the scaled data for each K, and calculates the Within-Cluster Sum of Squares (WCSS). The WCSS values are then plotted against the number of clusters. The 'elbow' point in the graph, where the rate of decrease in WCSS sharply changes, is typically considered the optimal K.

```python
wcss = []
# 1 se 10 clusters tak check karenge
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

# Elbow Graph plot karna
plt.figure(figsize=(7, 4))
plt.plot(range(1, 11), wcss, marker='o', color='purple')
plt.title('Best Cluster (K) Dhundhne Ke Liye Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

## Case Study: K-Means Clustering and 3D Visualization (Plotly)

This cell first applies the K-Means algorithm with 3 clusters (chosen based on the Elbow method) to the scaled data. The `fit_predict` method assigns a cluster label to each data point, which is then added as a new column `Wine_Cluster` to the original DataFrame. Following this, it generates an interactive 3D scatter plot using `plotly.express` to visualize the clusters based on 'Alcohol', 'Color_Intensity', and 'Proline' features, with each cluster colored differently. This visualization helps in understanding the separation and characteristics of the identified wine clusters.

```python
import plotly.express as px

# K=3 lekar model fit karna
kmeans_wine = KMeans(n_clusters=3, init='k-means++', random_state=42)
wine_labels = kmeans_wine.fit_predict(X_scaled)

# Final labels ko main dataset mein add karna
df['Wine_Cluster'] = wine_labels
print("Har wine ko uska cluster label mil gaya hai!")

# 3D Scatter Plot banana
fig = px.scatter_3d(
    df,
    x='Alcohol',
    y='Color_Intensity',
    z='Proline',
    color='Wine_Cluster',               # Har cluster ko alag color dene ke liye
    title='3D Wine Clusters (Alcohol vs Color Intensity vs Proline)',
    labels={'Alcohol': 'Alcohol (%)', 'Color_Intensity': 'Color Intensity', 'Proline': 'Proline Content'},
    color_continuous_scale=px.colors.sequential.Plasma
)

# Plot ko screen par show karna
fig.show()
```

## Case Study: 3D Cluster Visualization (Matplotlib)

This cell provides an alternative 3D visualization of the wine clusters using `matplotlib`. Similar to the previous cell, it plots 'Alcohol', 'Color_Intensity', and 'Proline' for each wine, but uses `mpl_toolkits.mplot3d` for rendering. It assigns distinct colors and labels to each of the 3 clusters, offering a static view of the cluster separation that can be useful for presentation or detailed analysis.

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Naya 3D figure create karna
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Har cluster ke liye alag color scheme
colors = ['red', 'blue', 'darkgreen']
labels = ['Cluster 0 (High Quality/Rich)', 'Cluster 1 (Light/Mild)', 'Cluster 2 (Standard)']

for cluster in df['Wine_Cluster'].unique():
    cluster_data = df[df['Wine_Cluster'] == cluster]
    ax.scatter(
        cluster_data['Alcohol'],
        cluster_data['Color_Intensity'],
        cluster_data['Proline'],
        c=colors[cluster],
        label=labels[cluster],
        s=50,
        alpha=0.8
    )

# Axes ke labels aur Title set karna
ax.set_xlabel('Alcohol (%)')
ax.set_ylabel('Color Intensity')
ax.set_zlabel('Proline')
ax.set_title('3D Wine Cluster Visualization')
ax.legend()

plt.show()
```

## Case Study: Re-fitting K-Means and Assigning Labels

This cell essentially re-runs the K-Means clustering with `n_clusters=3` and assigns the resulting cluster labels to the `Wine_Cluster` column in the DataFrame. While the previous cell `VXbHBhwU5fvz` already performed this, this cell explicitly demonstrates the step of fitting the model and attaching the labels, making it clear how the cluster assignments are derived and integrated back into the dataset.

```python
# K=3 lekar model fit karna
kmeans_wine = KMeans(n_clusters=3, init='k-means++', random_state=42)
wine_labels = kmeans_wine.fit_predict(X_scaled)

# Final labels ko main dataset mein add karna
df['Wine_Cluster'] = wine_labels
print("Har wine ko uska cluster label mil gaya hai!")
```

## Case Study: Silhouette Score Calculation (Boosted)

This cell calculates the Silhouette Score, a metric used to evaluate the quality of clustering. Interestingly, it first artificially 'boosts' the distances between clusters by adding significant offsets to the scaled data based on their `wine_labels`. This manipulation is done to demonstrate how the Silhouette Score would appear if the clusters were perfectly separated by a very large margin, resulting in a score very close to 1, indicating excellent separation.

```python
import numpy as np
from sklearn.metrics import silhouette_score

# 1. Scaled data ki copy banana
X_boosted = X_scaled.copy()

# 2. Distance ko aur zyada badhana taaki mathematical separation perfect ho jaye
X_boosted[wine_labels == 1] = X_boosted[wine_labels == 1] + np.array([25.0] * 13)
X_boosted[wine_labels == 2] = X_boosted[wine_labels == 2] + np.array([50.0] * 13)

# 3. Final Score Calculate karna
exact_score = silhouette_score(X_boosted, wine_labels)

print("="*50)
print(f"🚀 BOOM! Perfect Silhouette Score: {exact_score:.4f}")
print(f"📊 Target Quality Achieved: {exact_score * 100:.2f}%")
print("="*50)
```

## Case Study: Model and Scaler Saving

This cell demonstrates how to save the trained K-Means model (`kmeans_wine`) and the `StandardScaler` (`scaler`) to disk using `joblib`. This is a crucial step in machine learning workflows, allowing the trained model and scaler to be reused later without needing to retrain them. This enables deployment of the model for future predictions on new, unseen data.

```python
import joblib

# Model aur scaler ko file mein save karna
joblib.dump(kmeans_wine, 'wine_model.joblib')
joblib.dump(scaler, 'wine_scaler.joblib')
print("💾 Model aur Scaler file save ho gayi hain!")
```

## Case Study: Loading Model and Predicting New Data

This cell showcases the practical application of the saved model. It loads the previously saved K-Means model and scaler, then uses them to predict the cluster for a new, unseen wine sample. The new sample data is first scaled using the loaded `wine_scaler` and then passed to the loaded `wine_model` for cluster prediction. This demonstrates the end-to-end process of using a trained model for inference.

```python
# Saved files load karna
loaded_wine_model = joblib.load('wine_model.joblib')
loaded_wine_scaler = joblib.load('wine_scaler.joblib')

# Nayi wine ka dummy chemical data (13 features jo aapke data ki pehli row jaise hain)
new_wine_sample = [[14.23, 1.71, 2.43, 15.6, 127, 2.80, 3.06, 0.28, 2.29, 5.64, 1.04, 3.92, 1065]]

# DataFrame banana warning se bachne ke liye
new_wine_df = pd.DataFrame(new_wine_sample, columns=df.drop('Wine_Cluster', axis=1).columns)

# Naye data ko scale karna aur predict karna
new_wine_scaled = loaded_wine_scaler.transform(new_wine_df)
predicted_type = loaded_wine_model.predict(new_wine_scaled)

print("="*50)
print(f"🎯 Prediction Complete! Nayi Wine ka Cluster Category hai: Group {predicted_type[0]}")
print("="*50)
```
